```python
import pytesseract
from PIL import Image
text = pytesseract.image_to_string(Image.open('images\CV_NguyenNgocThang_DevOps.jpg'))
print(text)

# Lưu kết quả vào file ket_qua_ocr.txt trong cùng thư mục
with open('ket_qua_ocr.txt', 'w', encoding='utf-8') as f:
    f.write(text)

print("Đã trích xuất xong! Hãy mở file ket_qua_ocr.txt để xem toàn bộ nội dung.")
```

```python
import os
import fitz  # PyMuPDF
from pathlib import Path

data_dir = r'D:\OCR\dataset_cv\data'

for folder_name in os.listdir(data_dir):
    folder_path = os.path.join(data_dir, folder_name)
    if os.path.isdir(folder_path) and not folder_name.endswith('_jpg'):
        # Tạo thư mục mới với hậu tố _jpg
        new_folder_name = f"{folder_name}_jpg"
        new_folder_path = os.path.join(data_dir, new_folder_name)
        os.makedirs(new_folder_path, exist_ok=True)

        # Duyệt qua các file pdf
        for filename in os.listdir(folder_path):
            if filename.lower().endswith('.pdf'):
                pdf_path = os.path.join(folder_path, filename)
                try:
                    doc = fitz.open(pdf_path)
                    for page_num in range(len(doc)):
                        page = doc.load_page(page_num)
                        pix = page.get_pixmap(dpi=300)
                        
                        img_filename = f"{os.path.splitext(filename)[0]}_page{page_num+1}.jpg"
                        img_path = os.path.join(new_folder_path, img_filename)
                        pix.save(img_path)
                except Exception as e:
                    print(f"Lỗi file {pdf_path}: {e}")
        print(f"Đã xử lý xong {folder_name}")
print("Hoàn tất!")
```
